## Test 3: Multi-Channel with Patterns

```python
# Create patterned data to better visualize channel effects
pattern_data = np.zeros((3, 200, 200, 3), dtype=np.uint8)

# Frame 0: Red gradient
pattern_data[0, :, :, 0] = np.linspace(0, 255, 200).reshape(1, -1).repeat(200, axis=0).astype(np.uint8)

# Frame 1: Green gradient
pattern_data[1, :, :, 1] = np.linspace(0, 255, 200).reshape(-1, 1).repeat(200, axis=1).astype(np.uint8)

# Frame 2: Blue checkerboard
for i in range(0, 200, 20):
    for j in range(0, 200, 20):
        if (i + j) % 40 == 0:
            pattern_data[2, i:i+20, j:j+20, 2] = 255

pattern_src = THWCSequenceSource(pattern_data)
```

```python
# Display patterned source
pattern_src
```

### Verification Checklist for Test 3

- [ ] Frame 0 shows horizontal red gradient
- [ ] Frame 1 shows vertical green gradient
- [ ] Frame 2 shows blue checkerboard
- [ ] Toggling Channel 0 shows/hides red component
- [ ] Toggling Channel 1 shows/hides green component
- [ ] Toggling Channel 2 shows/hides blue component
- [ ] Unchecking all channels shows blank/black image
